# Experiment 11

```python
# Example constraints: x - y ≤ k
constraints = [
    ("x", "y", 3),   # x - y ≤ 3
    ("y", "z", 2),   # y - z ≤ 2
    ("z", "x", -10)  # z - x ≤ -10 (this may create conflict)
]
nodes = set()
edges = []

for u, v, w in constraints:
    nodes.add(u)
    nodes.add(v)
    edges.append((v, u, w))  # convert x - y ≤ k → y → x with weight k
```

```python
# Initialize distances
dist = {node: 0 for node in nodes}

# Relax edges
for _ in range(len(nodes) - 1):
    for u, v, w in edges:
        if dist[v] > dist[u] + w:
            dist[v] = dist[u] + w

# Check negative cycle
is_satisfiable = True
for u, v, w in edges:
    if dist[v] > dist[u] + w:
        is_satisfiable = False
        break

print("Satisfiable:", is_satisfiable)
```

```python
!pip install networkx matplotlib

import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

# Add edges (same transformation you used earlier)
for u, v, w in constraints:
    G.add_edge(v, u, weight=w)  # y → x with weight k
```

```python
pos = nx.spring_layout(G)

# Draw nodes
nx.draw(G, pos, with_labels=True, node_size=2000, font_size=12)

# Draw edge labels (weights)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.title("Difference Logic Constraint Graph")
plt.show()
```
